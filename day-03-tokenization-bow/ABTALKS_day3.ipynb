```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string

print("Setup done. NLTK ready.")
```

```python
paragraph = """Machines don't read words, they read numbers. Text preprocessing
is the first step in any NLP pipeline, turning raw human language into
structured data a model can actually learn from."""
tokens = word_tokenize(paragraph)
print("Original paragraph:")
print(paragraph)
print("\nTokens:")
print(tokens)
print(f"\nTotal tokens: {len(tokens)}")
```

```python
stop_words = set(stopwords.words('english'))
punctuation = set(string.punctuation)
filtered_tokens = [
    word.lower() for word in tokens
    if word.lower() not in stop_words and word not in punctuation
]
print("Before (raw tokens):")
print(tokens)
print(f"Count: {len(tokens)}")
print("\nAfter (stop words + punctuation removed):")
print(filtered_tokens)
print(f"Count: {len(filtered_tokens)}")
```

```python
from collections import Counter
bow = Counter(filtered_tokens)
print("Bag-of-Words representation:")
for word, count in bow.items():
    print(f"  {word}: {count}")
vocab = sorted(bow.keys())
bow_vector = [bow[word] for word in vocab]
print("\nVocabulary:", vocab)
print("BoW vector:", bow_vector)
```
