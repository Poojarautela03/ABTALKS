```python
!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer, util
import numpy as np

model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded.")
```

```python
sentences = [
    "Python is a popular programming language for beginners.",
    "Machine learning models require large amounts of data.",
    "The stock market saw a sharp decline today.",
    "She adopted a golden retriever puppy last week.",
    "Neural networks are inspired by the human brain.",
    "The chef prepared a delicious pasta dish.",
    "Cloud computing allows businesses to scale easily.",
    "He went for a morning run in the park.",
    "Data preprocessing is crucial before training a model.",
    "The football match ended in a dramatic penalty shootout.",
    "Artificial intelligence is transforming healthcare diagnostics.",
    "They watched a thrilling movie at the cinema.",
    "Deep learning uses multiple layers of neural networks.",
    "The garden was full of blooming roses in spring.",
    "APIs allow different software systems to communicate.",
    "The mountain hike offered a breathtaking sunset view.",
    "Natural language processing helps computers understand text.",
    "She baked chocolate chip cookies for the party.",
    "GPUs significantly speed up deep learning training.",
    "The airplane landed safely despite the storm."
]

embeddings = model.encode(sentences)
print(f"Embedded {len(sentences)} sentences. Shape: {embeddings.shape}")
```

```python
#Semantic search
def semantic_search(query,top_k=3):
  query_embedding=model.encode(query)
  scores=[]
  for i in range(len(sentences)):
    score=util.cos_sim(query_embedding,embeddings[i]).item()
    scores.append((score,sentences[i]))
  scores.sort(reverse=True,key=lambda x:x[0])
  print(f"Query:'{query}'\n")
  print("Top results (semantic search):")
  for i in range(top_k):
    score,sentence=scores[i]
    print(f"[{score:.4f}]{sentence}")

semantic_search("AI in medicine")
```

```python
#keyword search
def keyword_search(query,top_k=3):
  query_words=query.lower().split()
  scores=[]
  for sentence in sentences:
    sentence_words=sentence.lower().replace('.','').split()
    overlap=0
    for word in query_words:
      if word in sentence_words:
        overlap+=1
    scores.append((overlap,sentence))
  scores.sort(reverse=True,key=lambda x:x[0])
  print(f"Query:'{query}'\n")
  print("Top results (keyword search):")
  for i in range(top_k):
    overlap,sentence=scores[i]
    print(f"[{overlap}word overlap]{sentence}")
keyword_search("AI in medicine")
```

```python
test_queries = [
    "AI in medicine",
    "training a neural network",
    "outdoor activities"
]

for q in test_queries:
    print("=" * 60)
    semantic_search(q)
    print()
    keyword_search(q)
    print()
```
