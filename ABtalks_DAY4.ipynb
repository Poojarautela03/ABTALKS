```python
!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim

model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded.")
```

```python
sentences = [
    "The cat sat on the mat.",
    "A dog was resting on the rug.",
    "I love programming in Python.",
    "Machine learning models need data to train.",
    "The stock market crashed yesterday."
]
embeddings = model.encode(sentences)
print("Shape of embeddings:", embeddings.shape)
print("First embedding (first 10 values):", embeddings[0][:10])
```

```python
import itertools
print("Pairwise similarity scores:\n")
for i, j in itertools.combinations(range(len(sentences)), 2):
    score = cos_sim(embeddings[i], embeddings[j]).item()
    print(f"'{sentences[i]}'\n vs\n'{sentences[j]}'\n → similarity: {score:.4f}\n")
```

```python
pairs = []
for i, j in itertools.combinations(range(len(sentences)), 2):
    score = cos_sim(embeddings[i], embeddings[j]).item()
    pairs.append((score, sentences[i], sentences[j]))
pairs.sort(reverse=True)
print("Most similar pair:")
print(f"  {pairs[0][1]}\n  {pairs[0][2]}\n  Score: {pairs[0][0]:.4f}\n")
print("Least similar pair:")
print(f"  {pairs[-1][1]}\n  {pairs[-1][2]}\n  Score: {pairs[-1][0]:.4f}")
```
