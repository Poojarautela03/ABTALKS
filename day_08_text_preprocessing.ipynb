```python
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re

print("Setup done.")
```

```python
samples = {
    "social_media": "OMG this new phone is AMAZING!!! 😍🔥 Best purchase ever, y'all NEED to get one!!! #tech #obsessed",
    "news_article": "The government announced, on Tuesday, a new policy aimed at reducing carbon emissions by 40% over the next decade.",
    "customer_review": "I ordered this product last week and honestly... it's okay?? Not the BEST quality but does the job for the price. Would NOT recommend for heavy use!!"
}
for name,text in samples.items():
  print(f"[{name}]")
  print(text)
  print()
```

```python
for name,text in samples.items():
   print(f"=== {name} ===")
   word_tokens=word_tokenize(text)
   print(word_tokens)
   sent_tokens = sent_tokenize(text)
   print(f"\nSentence tokens ({len(sent_tokens)}):")
   print(sent_tokens)
   print()
```

```python
stop_words=set(stopwords.words('english'))
for name,text in samples.items():
    tokens=word_tokenize(text.lower())
    before_vocab = set(tokens)
    filtered = [w for w in tokens if w not in stop_words]
    after_vocab = set(filtered)

    print(f"=== {name} ===")
    print(f"Before removal ({len(before_vocab)} unique words): {before_vocab}")
    print(f"After removal ({len(after_vocab)} unique words): {after_vocab}")
    print(f"Vocabulary reduced by: {len(before_vocab) - len(after_vocab)} words\n")
```

```python
def strip_punctuation(text):
  return re.sub(r'[^\w\s]','',text)
for name, text in samples.items():
    cleaned = strip_punctuation(text)
    print(f"=== {name} ===")
    print(f"Before: {text}")
    print(f"After:  {cleaned}\n")
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for name, text in samples.items():
    tokens = word_tokenize(strip_punctuation(text).lower())
    filtered = [w for w in tokens if w not in stop_words]

    stemmed = [stemmer.stem(w) for w in filtered]
    lemmatized = [lemmatizer.lemmatize(w) for w in filtered]

    print(f"=== {name} ===")
    print(f"{'Original':<15}{'Stemmed':<15}{'Lemmatized':<15}")
    for orig, stem, lemma in zip(filtered, stemmed, lemmatized):
        print(f"{orig:<15}{stem:<15}{lemma:<15}")
    print()
```

## Why Each Preprocessing Step Matters

**Tokenization** breaks continuous text into discrete units (words/sentences) a model can process individually. Skip this, and the model sees one giant unstructured string with no concept of word boundaries — making any downstream counting, matching, or embedding impossible.

**Stopword removal** strips high-frequency, low-information words ("the," "is," "and"). Skipping this bloats vocabulary size and dilutes signal — in bag-of-words or TF-IDF models, stopwords dominate word counts without adding meaning, drowning out the words that actually distinguish one document from another.

**Punctuation removal** eliminates symbols that fragment tokens unnecessarily (e.g. "amazing!!!" vs "amazing" being treated as different words). Skipping this artificially inflates vocabulary size and creates sparse, redundant features that hurt model generalization.

**Stemming** (PorterStemmer) crudely chops words to a root form using rule-based suffix stripping — fast, but can produce non-words (e.g. "studies" → "studi"). **Lemmatization** (WordNetLemmatizer) uses a dictionary to return the true base form (e.g. "studies" → "study"), preserving meaning more accurately but running slower. Skipping normalization entirely means "run," "running," and "ran" are treated as three unrelated words, fragmenting the vocabulary and weakening the model's ability to recognize they refer to the same concept.

**Casing normalization** (lowercasing) ensures "Amazing" and "amazing" are treated as the same token. Skip this, and the model doubles its vocabulary for no semantic reason — every capitalized word at a sentence's start becomes a separate "unknown" token from its lowercase counterpart elsewhere in the text.

Together, these steps transform noisy, redundant, human-formatted text into a clean, consistent vocabulary — which is what makes accurate pattern-matching, classification, and embedding possible downstream.
