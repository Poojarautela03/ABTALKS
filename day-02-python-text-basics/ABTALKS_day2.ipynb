```python
def cleaned_text(text):
  text=text.lower()
  punctuation=".,!?;:\"'()[]{}-"
  cleaned=""
  for char in text:
    if char not in punctuation:
      cleaned+=char
  words=cleaned.split()
  cleaned=" ".join(words)
  return cleaned

def word_freq(text):
   cleaned=cleaned_text(text)
   words=cleaned.split()
   freq={}
   for word in words:
    if word in freq:
      freq[word]+=1
    else:
      freq[word]=1
   return freq


```

```python
sentence = "The quick brown fox jumps over the lazy dog. The dog barks!"
cleaned=cleaned_text(sentence)
print("Cleaned text:", cleaned)
freq=word_freq(sentence)
print("Word frequency:", freq)
for word, count in freq.items():
    print(word, ":", count)
```
